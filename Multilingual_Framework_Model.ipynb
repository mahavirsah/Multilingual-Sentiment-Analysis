```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification,TFBertModel
import tensorflow as tf
```

```python

def load_model(language):
    if language == 1:  # English
        tokenizer = AutoTokenizer.from_pretrained("/content/drive/MyDrive/English/englishfine_tuned_bert")
        model = AutoModelForSequenceClassification.from_pretrained("/content/drive/MyDrive/English/englishfine_tuned_bert")
        return model, tokenizer
    elif language == 2:  # Nepali
        # model = TFBertModel.from_pretrained("/content/drive/MyDrive/Nepali_Dataset/Nepali Sentiment Model")
        # tokenizer = AutoTokenizer.from_pretrained("/content/drive/MyDrive/Nepali_Dataset/Nepali Sentiment Model")

        tokenizer = AutoTokenizer.from_pretrained("/content/drive/MyDrive/Nepali_Dataset/nepali_finetuned")
        model = AutoModelForSequenceClassification.from_pretrained("/content/drive/MyDrive/Nepali_Dataset/nepali_finetuned")
        return model,tokenizer
    elif language == 3:  # Hindi
        tokenizer = AutoTokenizer.from_pretrained("/content/drive/MyDrive/Hindi_Dataset/hindi_finetuned_model")
        model = AutoModelForSequenceClassification.from_pretrained("/content/drive/MyDrive/Hindi_Dataset/hindi_finetuned_model")
        return model, tokenizer

    else:
        return None, None
```

```python
# Predict sentiment
# def predict_sentiment(model, tokenizer, text, language):
#     if language in [1, 2]:  # English or Nepali
#         inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)
#         outputs = model(**inputs)
#         sentiment = outputs.logits.argmax(dim=1).item()
#     elif language == 3:  # Hindi
#         prediction = model.predict([text])
#         sentiment = prediction.argmax()
#     else:
#         sentiment = None
#     return sentiment
def predict_sentiment(model, tokenizer, text, language):
    if language in [1, 2, 3]:  # English, Nepali or Hindi
        inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=128)
        outputs = model(**inputs)
        sentiment = outputs.logits.argmax(dim=1).item()
    else:
        sentiment = None
    return sentiment
```

```python
# Interpret sentiment results
def interpret_sentiment(sentiment, language):
    if language in [1, 2]:  # English or Nepali (2 classes: Positive, Negative)
        sentiment_map = {0: "Negative", 1: "Positive"}
    elif language == 3:  # Hindi (3 classes: Positive, Negative, Neutral)
        sentiment_map = {0: "Negative", 1: "Neutral", 2: "Positive"}
    else:
        sentiment_map = {}
    return sentiment_map.get(sentiment, "Unknown")
```

```python

def main():
    while True:
        print("Enter text for sentiment analysis:")
        text = input("> ")

        print("Select language:\n1. English\n2. Nepali\n3. Hindi")
        try:
            language = int(input("> "))
        except ValueError:
            print("Invalid input. Please enter 1, 2, or 3.")
            continue

        if language not in [1, 2, 3]:
            print("Invalid choice. Please select a valid language.")
            continue
        print("Loading model...")
        model = load_model(language)

        if model is None:
            print("Failed to load model. Exiting.")
            break1
        tokenizer = None
        if language in [1, 2]:
            model, tokenizer = model

        print("Predicting sentiment...")
        sentiment = predict_sentiment(model, tokenizer, text, language)

        # Interpret sentiment result
        sentiment_result = interpret_sentiment(sentiment, language)
        print(f"Sentiment: {sentiment_result}")

        # Ask if the user wants to run again
        print("Do you want to analyze another text? (yes/no)")
        again = input("> ").strip().lower()
        if again != "yes":
            print("Exiting the program. Goodbye!")
            break

if __name__ == "__main__":
    main()
```
